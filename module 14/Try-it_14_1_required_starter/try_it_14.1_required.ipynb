We provide a small dataset with health data where your goal is to predict whether or not the individuals survive.

```python
import pandas as pd
import plotly.express as px
from sklearn.model_selection import (
    train_test_split,
    GridSearchCV,
    RandomizedSearchCV,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier
from skopt import BayesSearchCV
```

```python
data = pd.read_csv('data/Whickham.txt')
```

```python
data.head()
```

```python
X = data[['smoker', 'age']]
y = data['outcome']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
params = {
    "tree__max_depth": [1, 2, 3, 4, 5, 6, 7],
    "tree__min_samples_split": [2, 3, 4, 5, 6],
    "tree__criterion": ["gini", "entropy", "log_loss"],
    "tree__min_samples_leaf": [1, 2, 3, 4, 5, 6, 7],
}
```

```python
pipe = Pipeline(
    [
        ("onehot", OneHotEncoder()),
        ("scaler", StandardScaler(with_mean=False)),
        ("tree", DecisionTreeClassifier()),
    ]
)
```

```python
gs = GridSearchCV(pipe, param_grid=params, verbose=2).fit(X_train, y_train)
```

```python
rs = RandomizedSearchCV(pipe, param_distributions=params, verbose=2).fit(X_train, y_train)
```

```python
bayes = BayesSearchCV(pipe, search_spaces=params, verbose=2).fit(X_train, y_train)
```

```python
gs.best_params_
```

```python
rs.best_params_
```

```python
bayes.best_params_
```

```python
gs_time_data = {
    'mean_fit_time': gs.cv_results_['mean_fit_time'],
    'mean_score_time': gs.cv_results_['mean_score_time'],
    'std_score_time': gs.cv_results_['std_score_time'],
}

gs_time_data_df = pd.DataFrame(gs_time_data)
```

```python
rs_time_data = {
    'mean_fit_time': rs.cv_results_['mean_fit_time'],
    'mean_score_time': rs.cv_results_['mean_score_time'],
    'std_score_time': rs.cv_results_['std_score_time'],
}

rs_time_data_df = pd.DataFrame(rs_time_data)
```

```python
bayes_time_data = {
    'mean_fit_time': bayes.cv_results_['mean_fit_time'],
    'mean_score_time': bayes.cv_results_['mean_score_time'],
    'std_score_time': bayes.cv_results_['std_score_time'],
}

bayes_time_data_df = pd.DataFrame(bayes_time_data)
```

```python
fig = px.line(gs_time_data_df, labels={'value': 'Seconds', 'index': 'Fit'},
              title='GridSearchCV takes the longs and fits all possibilities')
fig.write_image('images/gs_time.png')
```

```python
fig = px.line(rs_time_data_df, labels={'value': 'Seconds', 'index': 'Fit'},
              title='RandomSearchCV has less fits and takes least amount of time')
fig.write_image('images/rs_time.png')
```

```python
fig = px.line(bayes_time_data_df, labels={'value': 'Seconds', 'index': 'Fit'},
              title='BayesSearchCV has less fits than GridSearchCV but run time was slower')
fig.write_image('images/bayes_time.png')
```
